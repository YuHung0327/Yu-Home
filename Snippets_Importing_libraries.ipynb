```
import sqlite3
import pandas as pd

# 連接資料庫
conn = sqlite3.connect('my_cloud_database.db')

# 1. 列出資料庫中所有的資料表
tables_query = "SELECT name FROM sqlite_master WHERE type='table';"
tables = pd.read_sql_query(tables_query, conn)
print("目前資料庫中的資料表：")
display(tables)
```

```
# 2. 針對 'housing_data' 進行簡單的數據分析
# 計算各個屋齡 (housing_median_age) 的平均房價與平均收入
analysis_query = """
SELECT
    housing_median_age,
    AVG(median_income) as avg_income,
    AVG(median_house_value) as avg_house_value
FROM housing_data
GROUP BY housing_median_age
ORDER BY housing_median_age
"""
analysis_df = pd.read_sql_query(analysis_query, conn)

print("依屋齡分類的平均收入與房價：")
display(analysis_df.head(10))

conn.close()
```

### 方案一：使用 SQLite (輕量級，儲存在 Colab 空間)
這不需要額外安裝，適合快速開發與測試。

```
import sqlite3
import pandas as pd

# 建立或連接資料庫檔案
conn = sqlite3.connect('my_cloud_database.db')
cursor = conn.cursor()

# 建立資料表
cursor.execute('''
CREATE TABLE IF NOT EXISTS projects (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    status TEXT
)
''')

# 插入測試資料
cursor.execute("INSERT INTO projects (name, status) VALUES ('Cloud DB Setup', 'In Progress')")
conn.commit()

# 查詢並顯示結果
query_df = pd.read_sql_query("SELECT * FROM projects", conn)
display(query_df)
conn.close()
```
